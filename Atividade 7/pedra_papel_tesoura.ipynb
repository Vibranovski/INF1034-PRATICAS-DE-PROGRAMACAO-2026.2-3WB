```python
import random
from typing import Dict, Tuple

ROCK = "r"
PAPER = "p"
SCISSORS = "s"

emojis: Dict[str, str] = {
    ROCK: "✊",  # Rock
    PAPER: "✋",  # Paper
    SCISSORS: "✌️"   # Scissors
}

rules: Dict[str, Tuple[str]] = {
    ROCK: (SCISSORS,),  # Rock beats scissors
    PAPER: (ROCK,),     # Paper beats rock
    SCISSORS: (PAPER,)  # Scissors beats paper
}

choices = tuple(emojis.keys())

def get_user_choice() -> str:
    """Prompt the user for their choice and validate it."""
    while True:
        user_choice = input("Rock, Paper, Scissors? (r/p/s): ").lower().strip()
        if user_choice in choices:
            return user_choice
        print("Invalid choice! Please enter 'r', 'p', or 's'.")

def displaychoices(user_choice: str, computer_choice: str) -> None:
    """Display the choices with emojis."""
    print(f"You chose: {emojis[user_choice]}")
    print(f"Computer chose: {emojis[computer_choice]}")

def determinate_winner(user_choice: str, computer_choice: str) -> None:
    """Determine and display the winner."""
    if user_choice == computer_choice:
        print("It's a tie!")
    elif computer_choice in rules[user_choice]:
        print("You win!")
    else:
        print("You lose!")

def play_again() -> bool:
    """Ask the user if they want to play again."""
    return input("Play again? (y/n): ").lower().strip() == 'y'

def play_game() -> None:
    """Main game loop."""
    print("Welcome to Rock, Paper, Scissors!")
    while True:
        user_choice = get_user_choice()
        computer_choice = random.choice(choices)
        displaychoices(user_choice, computer_choice)
        determinate_winner(user_choice, computer_choice)
        if not play_again():
            print("Thanks for playing!")
            break

if __name__ == "__main__":
    play_game()
```

```python
# Função principal conforme solicitado no desafio
def pedra_papel_tesoura(jogador1: str, jogador2: str) -> str:
    # Normaliza os valores para letras minúsculas e remove espaços
    jogador1 = jogador1.strip().lower()
    jogador2 = jogador2.strip().lower()

    # Lista de escolhas válidas
    opcoes_validas = ("pedra", "papel", "tesoura")

    # Verifica se ambos os jogadores fizeram escolhas válidas
    if jogador1 not in opcoes_validas or jogador2 not in opcoes_validas:
        return "Escolha pedra, papel ou tesoura para jogar"

    # Caso de empate
    if jogador1 == jogador2:
        return "empate"

    # Define as regras do jogo: quem ganha de quem
    regras = {
        "pedra": "tesoura",   # Pedra ganha da tesoura
        "papel": "pedra",     # Papel ganha da pedra
        "tesoura": "papel"    # Tesoura ganha do papel
    }

    # Verifica quem ganhou
    if regras[jogador1] == jogador2:
        return "jogador 01 ganhou"
    else:
        return "jogador 02 ganhou"
    
print(pedra_papel_tesoura("pedra", "papel"))
print(pedra_papel_tesoura("tesoura", "papel"))
print(pedra_papel_tesoura("papel", "papel"))
print(pedra_papel_tesoura("chave", "pedra"))
```
